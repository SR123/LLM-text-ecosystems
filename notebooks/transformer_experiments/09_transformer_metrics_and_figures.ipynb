This notebook is analysis-only: it uses saved artifacts to rebuild figures/tables without retraining.

```python
from pathlib import Path
import json
import sys

import pandas as pd

NB_DIR = Path.cwd().resolve()
ROOT = NB_DIR.parents[2] if NB_DIR.name == "active" else Path.cwd().resolve()
SRC = ROOT / "GitHub" / "src"
if str(SRC) not in sys.path:
    sys.path.insert(0, str(SRC))

from drift_selection.plots import plot_teacher_student_bar_chart, plot_environment_summary_table
```

```python
base = ROOT / "GitHub/data/outputs/transformer_conan_doyle"
metrics_csv = base / "evaluation/csv/teacher_student_metrics_pilot.csv"
manifest_path = base / "environments/environment_generation_manifest_pilot.json"

metrics_df = pd.read_csv(metrics_csv)
display(metrics_df)

plot_teacher_student_bar_chart(
    metrics_df,
    ROOT / "GitHub/figures/paper_main/transformer_teacher_student_bars_from_notebook",
    title="Teacher/Student Policy Metrics (pilot)",
)
```

```python
env_manifest = json.loads(manifest_path.read_text())
rows = []
for name in ["original", "neutral", "selected"]:
    m = env_manifest[name]
    rows.append({"environment": name, "tokens": m.get("token_count", 0), "token_ids_path": m.get("token_ids_path", "")})
summary_df = pd.DataFrame(rows)
display(summary_df)

plot_environment_summary_table(summary_df, ROOT / "GitHub/figures/appendix/environment_summary_table")
```
