# 07 Transformer Conan Doyle Teacher/Student

This notebook runs the end-to-end tiny-transformer environmental distillation pilot:

1. tokenizer + corpus encoding
2. teacher training
3. original/neutral/selected environment generation
4. student training
5. policy/branch evaluation + figures

All paths are repository-relative.

```python
from pathlib import Path
import subprocess
import sys

NB_DIR = Path.cwd().resolve()
if NB_DIR.name == "active":
    ROOT = NB_DIR.parents[2]
else:
    ROOT = Path.cwd().resolve()

CONFIG = "GitHub/configs/transformer_teacher_student.yaml"
SCRIPTS = ROOT / "GitHub" / "scripts"
print("ROOT:", ROOT)
print("SCRIPTS:", SCRIPTS)
```

## Stage 1: Train or Load Teacher

```python
cmd = [sys.executable, str(SCRIPTS / "run_transformer_teacher_train.py"), "--root", str(ROOT), "--config", CONFIG]
subprocess.run(cmd, check=True)
```

## Stage 2: Generate Original / Neutral / Selected Environments (Pilot 100k)

```python
cmd = [sys.executable, str(SCRIPTS / "run_transformer_generate_environments.py"), "--root", str(ROOT), "--config", CONFIG, "--mode", "pilot"]
subprocess.run(cmd, check=True)
```

## Stage 3: Train Three Students

```python
cmd = [sys.executable, str(SCRIPTS / "run_transformer_student_train.py"), "--root", str(ROOT), "--config", CONFIG, "--mode", "pilot"]
subprocess.run(cmd, check=True)
```

## Stage 4: Evaluate + Build Figures

```python
cmd = [sys.executable, str(SCRIPTS / "run_transformer_evaluation.py"), "--root", str(ROOT), "--config", CONFIG, "--mode", "pilot"]
subprocess.run(cmd, check=True)
```

## Stage 5: Inspect Result Artifacts

```python
import pandas as pd
metrics_csv = ROOT / "GitHub/data/outputs/transformer_conan_doyle/evaluation/csv/teacher_student_metrics_pilot.csv"
manifest = ROOT / "GitHub/data/outputs/transformer_conan_doyle/evaluation/evaluation_manifest_pilot.json"
print("metrics:", metrics_csv)
print("manifest:", manifest)
if metrics_csv.exists():
    display(pd.read_csv(metrics_csv))
```
